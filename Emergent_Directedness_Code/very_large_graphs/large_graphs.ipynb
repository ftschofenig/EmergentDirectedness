```python
import pandas as pd
from joblib import load
import seaborn as sns
import matplotlib.pyplot as plt
import networkx as nx
```

```python
# Load the DataFrame from the joblib file
df = load('large_CPL.joblib')
```

```python
#filter to only include rows where Spreading_Density is larger than 0.1
df = df[df['Spreading_Density'] >= 0.01]
```

```python
# Map node counts to full parameter strings from large_graphs.py
param_map = {5000: 'n=5000, m=5, p=0.4',
             10000: 'n=10000, m=5, p=0.4', 20000: 'n=20000, m=5, p=0.4'}
df['Network'] = df['number_of_nodes'].map(param_map)

# Create a figure with two subplots side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharey=True)

# Left subplot: T < 1 (relative thresholds θ)
sns.lineplot(data=df[df['T'] < 1], x='T', y='Symmetry', hue='Network', marker=None, linewidth=1, palette='viridis', ax=axes[0])
axes[0].set_xlabel('θ', fontsize=14)
axes[0].set_ylabel(r'$\Xi_s$(RCS)', fontsize=14)
axes[0].set_title(r'$\Xi_s$(RCS) vs θ (relative thresholds) for Different Network Sizes')
axes[0].legend(title='Network Parameters')

# Right subplot: T >= 1 (absolute thresholds)
sns.lineplot(data=df[df['T'] >= 1], x='T', y='Symmetry', hue='Network', marker=None, linewidth=1, palette='viridis', ax=axes[1])
axes[1].set_xlabel('T', fontsize=14)
axes[1].set_ylabel(r'$\Xi_s$(RCS)', fontsize=14)
axes[1].set_title(r'$\Xi_s$(RCS) vs T (absolute thresholds) for Different Network Sizes')
axes[1].set_xticks(range(1, 6))
axes[1].legend(title='Network Parameters')
#axes[1].set_xlim(2,4)

# Adjust layout for better spacing
plt.tight_layout()

# Save the plot
plt.savefig('very_large_graphs_symmetry.png')
```
